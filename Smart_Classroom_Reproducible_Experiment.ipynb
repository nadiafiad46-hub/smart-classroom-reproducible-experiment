```python

import pandas as pd

# Example measurements (kWh)
data = {
    "Before": [10, 12, 8, 10],
    "After": [7, 9, 6, 8]
}

df = pd.DataFrame(data)

before_avg = df["Before"].mean()
after_avg = df["After"].mean()

saving = ((before_avg - after_avg) / before_avg) * 100

print(df)
print("Average Before:", before_avg, "kWh")
print("Average After:", after_avg, "kWh")
print("Energy Saving:", round(saving, 2), "%")
```

```python

# Reproducibility Test

original_saving = saving

# Recalculate using the same data
new_before_avg = df["Before"].mean()
new_after_avg = df["After"].mean()

reproduced_saving = (
    (new_before_avg - new_after_avg) / new_before_avg
) * 100

print("Original Result:", round(original_saving, 2), "%")
print("Reproduced Result:", round(reproduced_saving, 2), "%")

if abs(original_saving - reproduced_saving) < 1e-9:
    print("SUCCESS: Result Reproduced!")
else:
    print("WARNING: Results Do Not Match!")
```

```python

# Test the effect of changing measurements

modified_df = df.copy()

# Change one measurement
modified_df.loc[0, "After"] = 9

new_before = modified_df["Before"].mean()
new_after = modified_df["After"].mean()

new_saving = ((new_before - new_after) / new_before) * 100

print("Original Saving:", round(saving, 2), "%")
print("Modified Saving:", round(new_saving, 2), "%")
print("Difference:", round(new_saving - saving, 2), "percentage points")
```

```python

# Save the original measurements

df.to_csv("energy_measurements.csv", index=False)

print("Original dataset saved successfully!")
```

```python
import pandas as pd

# Load the saved dataset
reloaded_df = pd.read_csv("energy_measurements.csv")

# Recalculate energy saving
before = reloaded_df["Before"].mean()
after = reloaded_df["After"].mean()

reproduced_saving = ((before - after) / before) * 100

print("Dataset loaded successfully!")
print("Reproduced Energy Saving:", round(reproduced_saving, 2), "%")
```
